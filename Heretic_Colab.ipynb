# Heretic on Android — cloud GPU guide

Use your Android phone as the controller. Heretic runs in a hosted GPU runtime; this notebook does **not** run model modification on the phone itself. It modifies open-weight models, not ChatGPT or GPT-6.

## 1. Enable a GPU

In Colab choose **Runtime → Change runtime type → GPU**, save, then run the next cell. GPU availability and quotas depend on your account.

```python
import torch
if not torch.cuda.is_available():
    raise RuntimeError('No GPU detected. Choose Runtime → Change runtime type → GPU, then rerun.')
print('GPU:', torch.cuda.get_device_name(0))
!nvidia-smi
```

## 3. Select a model

Choose a supported Hugging Face model small enough for the assigned GPU. The example may still exceed some runtimes' memory. Do not paste access tokens into this notebook.

```python
MODEL_ID = 'Qwen/Qwen3-4B-Instruct-2507'
print('Model:', MODEL_ID)
```

## 4. Run Heretic

The default configuration performs refusal suppression. This can take a long time and use substantial GPU memory and disk. If it runs out of memory, stop and select a smaller supported model or consult the official configuration documentation.

```python
!heretic "$MODEL_ID"
```

## 5. Save your output

Runtime storage is temporary. Use Heretic's save/upload options or copy the finished model to persistent storage before disconnecting.
